```python
data = [
    {'po_number': 'PO1001',
     'part_number': 'A100',
     'carton_number': 'CT001',
     'expected_qty': 10,
     'received_qty': 10
    },
    {'po_number': 'PO1001',
     'part_number': 'A100',
     'carton_number': 'CT002',
     'expected_qty': 12,
     'received_qty': 9
    },
    {'po_number': 'PO1001',
     'part_number': 'B200',
     'carton_number': 'CT003',
     'expected_qty': 8,
     'received_qty': 10
    }
]

receiving_log = []

print(data)
```

```python
total_boxes = 0
normal_boxes = 0
short_boxes = 0
over_boxes = 0
total_short_qty = 0
total_over_qty = 0

for item in data:
    total_boxes += 1

    if item['expected_qty'] == item['received_qty']:
        print(f"{item['carton_number']}: OK")
        normal_boxes += 1

    elif item['received_qty'] < item['expected_qty']:
        diff = item['expected_qty'] - item['received_qty']
        print(f"{item['carton_number']}: SHORT by {diff}")
        short_boxes += 1
        total_short_qty += diff
    else:
        diff = item['received_qty'] - item['expected_qty']
        print(f"{item['carton_number']}: OVER by {diff}")
        over_boxes += 1
        total_over_qty += diff

print(f"盘点汇总报告")
print(f"检查箱数   {total_boxes}")
print(f"正常箱数   {normal_boxes}")
print(f"短装箱数   {short_boxes}")
print(f"超收箱数   {over_boxes}")

print(f"短装总件数  {total_short_qty}件")
print(f"超收总件数  {total_over_qty}件")
```

```python
while True:
    carton = input("Enter box number (enter DONE to exit): ")
    if carton == 'DONE':
        break

    while True:
        received_qty = input("Enter unit count: ")

        # 检查输入是否为整数
        try:
            received_qty = int(received_qty)
        except ValueError:
            print("输入格式错误，请输入整数格式！")
            continue

        # 检查输入是否为负数
        if received_qty < 0:
            print("输入不能为负数")
            continue

        break

    record = {
        'carton_number': carton,
        'received_qty': received_qty
    }
    receiving_log.append(record)

print(receiving_log)
```

```python
total_boxes = 0
normal_boxes = 0
short_boxes = 0
over_boxes = 0
not_received_boxes = 0
total_short_qty = 0
total_over_qty = 0

for item in data:
    found = False
    for log in receiving_log:
        if log['carton_number'] == item['carton_number']:
            found = True
            if log['received_qty'] == item['expected_qty']:
                print(f"{log['carton_number']}: 收货正常")
                normal_boxes += 1
            elif log['received_qty'] < item['expected_qty']:
                diff = item['expected_qty'] - log['received_qty']
                print(f"{log['carton_number']}: 短装{diff}件")
                short_boxes += 1
                total_short_qty += diff
            else:
                diff = log['received_qty'] - item['expected_qty']
                print(f"{log['carton_number']}: 超收{diff}件")
                over_boxes += 1
                total_over_qty += diff
            break
    if not found:
        print(f"{item['carton_number']}: NOT RECEIVED")
        not_received_boxes += 1

    total_boxes += 1

in_full_rate = normal_boxes / total_boxes * 100

print('=======收货汇总报告=======')
print(f"工厂清单总箱数: {len(data)}")
print(f"正常箱数: {normal_boxes}")
print(f"短装箱数: {short_boxes}")
print(f"超收箱数: {over_boxes}")
print(f"未收到箱数: {not_received_boxes}")
print('===================')
print(f"短缺总件数: {total_short_qty}")
print(f"超收总件数: {total_over_qty}")
print(f"收货准确率： {in_full_rate:.2f}%")
```
